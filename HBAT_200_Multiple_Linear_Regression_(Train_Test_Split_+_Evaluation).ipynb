## Step 2: Understand Multiple Linear Regression

Multiple linear regression means:

One dependent variable is predicted from multiple independent variables.

**Customer Satisfaction** from variables such as:

- Product Quality
- Technical Support
- Complaint Resolution
- Delivery Speed
- Salesforce Image

Conceptually:

$$ Y = \beta_0+ \beta_1X_1+ \beta_2X_2+ \beta_3X_3+ \cdots+ \epsilon $$

Where:

- \(Y\) = outcome
- \(X_1, X_2, X_3\) = predictors
- \(\beta_0\) = intercept
- \(\beta_1,\beta_2,\beta_3\) = regression coefficients
- \(\epsilon\) = unexplained error

This is what distinguishes multiple regression from a simple regression example like:

Year -> Income

(one X)

HBAT:

X1, X2, X3, X4, X5 -> Y

Multiple X variables.

## Step 3: Select Predictors (X) and Outcome (y)

For this practice we will predict **Customer Satisfaction (X19)** using five predictors: Product Quality (X6), Technical Support (X8), Complaint Resolution (X9), Delivery Speed (X18), and Salesforce Image (X12).

```python
# The answer we want the model to learn
y = df['X19']

# The information we give the model
X = df[
    [
        'X6',   # Product Quality
        'X8',   # Technical Support
        'X9',   # Complaint Resolution
        'X18',  # Delivery Speed
        'X12'   # Salesforce Image
    ]
]

# Features -> Target
```

## Step 4: Train/Test Split

**Before**

- `X`
- `y`

**After**

- `X_train`, `X_test`
- `y_train`, `y_test`

| Split | Share | Purpose |
|---|---|---|
| Train | 80% | Model learns here |
| Test | 20% | Final exam here |

The model is allowed to see `X_train` and `y_train`, but it does **not** learn from `X_test` and `y_test`.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

### Understand the four split objects

Conceptually:

- `X_train` = Predictor values used for learning.
- `y_train` = Correct answers corresponding to `X_train`.
- `X_test` = New predictor values the trained model hasn't learned from.
- `y_test` = Actual answers we compare the model's predictions against.

So:

`X_train -> model -> y_train`

then:

`X_test -> model -> predictions`

finally:

`predictions <-> y_test`

That is the whole logic of supervised machine learning.

## Step 5: Initialize the Model

```python
# Initialize the model
model = LinearRegression()
```

At this moment:

- The model knows nothing about HBAT.
- It is essentially an empty mathematical structure.

## Step 6: Train (Fit) the Model

```python
# Train the model
model.fit(X_train, y_train)
```

This is one of the most important lines in machine learning.

You are effectively telling Python:

*Study how these X variables relate to these known y values and estimate the regression equation.*

After `.fit()`, the model contains learned coefficients.

## Step 7: Examine the Intercept and Coefficients

```python
# Examine the intercept
model.intercept_
```

This represents \(\beta_0\) in the regression equation.

Suppose Python returns:

`1.35`

Then the equation begins: `Y = 1.35 + ...`

```python
# Examine the coefficients
model.coef_
```

```python
# Alternatively, display them in a labeled table
coefficients = pd.DataFrame({
    'Variable': X.columns,
    'Coefficient': model.coef_
})

coefficients
```

You might hypothetically obtain:

| Predictor | Coefficient |
|---|---|
| Product Quality | 0.42 |
| Technical Support | 0.21 |
| Complaint Resolution | 0.31 |
| Delivery Speed | 0.08 |
| Salesforce Image | 0.17 |

Then the equation becomes approximately:

`Y = 1.35 + 0.42*X6 + 0.21*X8 + 0.31*X9 + 0.08*X18 + 0.17*X12`

### Interpreting a coefficient correctly

Suppose Product Quality coefficient = 0.42.

The proper interpretation is:

*Holding the other predictor variables constant, a one-unit increase in Product Quality is associated with an estimated 0.42-unit increase in Customer Satisfaction.*

The phrase **holding the other variables constant** is essential in multiple regression, since each coefficient represents the unique contribution of that predictor after accounting for the others.

## Step 8: Make Predictions

```python
# Make predictions
y_pred = model.predict(X_test)
```

Notice what goes inside `.predict()`: `X_test`, **not** `y_test`.

Why? Because `y_test` contains the answers. The model receives the questions (`X_test`) and produces `y_pred`.

```python
# Compare predictions against reality
results = pd.DataFrame({
    'Actual': y_test,
    'Predicted': y_pred
})

results.head(10)
```

## Step 9: Evaluate the Model

Now we check how well the model performed on the test set it never saw during training.

| Metric | Meaning |
|---|---|
| MAE (Mean Absolute Error) | Average size of the prediction error, in the original units of `X19` |
| MSE (Mean Squared Error) | Average of the squared errors - penalizes big misses more |
| RMSE (Root Mean Squared Error) | Square root of MSE, back in the original units - easier to interpret than MSE |
| R-squared | Proportion of the variance in `X19` explained by the predictors (0 to 1, higher is better) |

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f'MAE  (Mean Absolute Error):      {mae:.3f}')
print(f'MSE  (Mean Squared Error):       {mse:.3f}')
print(f'RMSE (Root Mean Squared Error):  {rmse:.3f}')
print(f'R2   (R-squared):                {r2:.3f}')
```

**How to read these numbers:**

- MAE and RMSE are in the same units as Customer Satisfaction (X19). Smaller values mean the model's predictions land closer to the real answers.
- RMSE will always be greater than or equal to MAE; a big gap between them means the model has a few large misses.
- R-squared close to 1 means the five predictors explain most of the variation in satisfaction; R-squared close to 0 means they explain very little.

## Step 10: Visualize Actual vs. Predicted

```python
plt.figure(figsize=(6, 6))
plt.scatter(y_test, y_pred, alpha=0.7, edgecolor='k')

# Perfect-prediction reference line
lims = [min(y_test.min(), y_pred.min()), max(y_test.max(), y_pred.max())]
plt.plot(lims, lims, 'r--', label='Perfect prediction')

plt.xlabel('Actual Satisfaction (X19)')
plt.ylabel('Predicted Satisfaction')
plt.title('Actual vs. Predicted Customer Satisfaction')
plt.legend()
plt.show()
```

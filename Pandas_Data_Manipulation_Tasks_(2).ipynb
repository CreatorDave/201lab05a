```python
import pandas as pd
import numpy as np

# Sample sales data
df_sales = pd.DataFrame({
    'Store': ['A', 'B', 'A', 'C', 'B', 'C', 'A'],
    'Product': ['Apples', 'Oranges', 'Bananas', 'Apples', 'Bananas', 'Oranges', 'Bananas'],
    'Units_Sold': [30, 45, 20, 25, 30, 40, 50],
    'Price': [1.0, 0.8, 0.5, 1.2, 0.55, 0.75, 0.6],
    'Date': pd.to_datetime([
        '2024-01-01', '2024-01-02', '2024-01-03', '2024-01-01',
        '2024-01-02', '2024-01-03', '2024-01-04'
    ])
})

df_stores = pd.DataFrame({
    'Store': ['A', 'B', 'C'],
    'City': ['Prague', 'Brno', 'Ostrava'],
    'Manager': ['Alice', 'Bob', 'Charlie']
})
```

### Task 1: Preview the first 2 few rows of both DataFrames


```python
# Your code here
df_sales.head(2)
df_stores.head(2)
print(df_sales)
print(df_stores)
```

### Task 2: Create a new column `Revenue` as Units_Sold * Price

```python
# Your code here
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
print(df_sales)
```

### Task 3: Filter the rows to include only sales of Bananas

```python
# Your code here
df_sales[df_sales['Product'] == 'Bananas']
```

### Task 4: Filter the rows for sales after January 2, 2024

```python
# Your code here
df_sales[df_sales['Date'] > '2024-01-02']
```

### Task 5: Group by `Product` and calculate total units sold

```python
# Your code here
df_sales.groupby('Product')['Units_Sold'].sum()
```

### Task 6: Group by `Store` and `Product`, and show total revenue

```python
# Your code here
df_sales.groupby(['Store', 'Product'])['Revenue'].sum()
```

### Task 7: Merge `df_sales` with `df_stores` on the `Store` column

```python
# Your code here
df_sales.merge(df_stores, on='Store')
```

### Task 8: Sort the sales by `Revenue` in descending order

```python
# Your code here
df_sales.sort_values(by='Revenue', ascending=False)
```

### Task 9: Introduce a missing value and fill it with the mean

```python
# Your code here
df_sales.loc[0, 'Units_Sold'] = np.nan
df_sales.fillna(df_sales['Units_Sold'].mean())
```

### Task 10: Create a new boolean column `High_Seller` for Units_Sold > 30

```python
# Your code here
df_sales['High_Seller'] = df_sales['Units_Sold'] > 30
print(df_sales)
```

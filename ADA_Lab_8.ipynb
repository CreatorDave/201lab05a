1. Find the probability of getting a numbered card when a card is drawn from the pack of 52 cards.

In a standard 52-card deck, numbered cards are 2 through 10.
9 numbered ranks
4 suits per rank
Numbered cards: 9 X 4 = 36

Mathematically:

P(numbered card) = 36/52 = 9/13 ~ 0.6923

So the probability is approximately 69.23%

```python
total_cards = 52
numbered_cards = 9 * 4  # Cards 2 through 10 in four suit

probability_numbered_card = numbered_cards / total_cards

print("Probability", probability_numbered_card)
print(f"Percentage: {probability_numbered_card * 100:.2f}%")

```

2. You have a bag containing 8 red balls, 5 green balls, and 3 blue balls. If you randomly draw two balls without replacement, what is the probability that both balls are green?

There are 16 balls total:

8 + 5 + 3 + 16

The probability that the first ball is green is:

5/16

Without replacement, 4 green balls remain out of 15 total balls:

P(both green) = 5/16 X 4/15 = 20/240 = 1/12
P(both green) ~ 0.0833= 8.33%


```python
total_balls = 16
green_balls = 5



probability_both_green = (
    green_balls / total_balls
) * (
    (green_balls - 1) / (total_balls - 1)
)

print("Probability", probability_both_green)
print(f"Percentage: {probability_both_green * 100:.2f}%")
```

3. You roll a six-sided die. What is the probability of rolling an even number or a number greater than 4?

A six-sided die has outcomes:
{1,2,3,4,5,6}

Even numbers:
{2,4,6}

Numbers greater than 4:
{5,6}

The union is:
{2,4,5,6}

So there are 4 favorable outcomes out of 6:

P(even or greater than 4) = 4/6 = 2/3 = 66.67%

The outcome 6 is counted only once because it satisfies both conditions.

```python
di_total = 6
condition = 4

probability_even_or_greater_than_4 = condition / di_total

print("Probability", probability_even_or_greater_than_4)
print(f"Percentage: {probability_even_or_greater_than_4 * 100:.2f}%")
```

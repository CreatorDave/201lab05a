## Section 2: Data Preparation

**Task 2.1:** Convert `Order Date` and `Ship Date` columns to datetime format.

```python
# Your code here
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
print(df.dtypes)
```

**Task 2.2:** Create a new column called `Delivery Time (Days)` that calculates the number of days between the order date and ship date.

```python
# Your code here
df['Delivery Time (Days)'] = (df['Ship Date'] - df['Order Date']).dt.days
df.head()
```

## Section 3: Exploratory Data Analysis

**Task 3.1:** Calculate the total sales for each product category. Sort the results from highest to lowest.

```python
# Your code here
df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
```

**Task 3.2:** Calculate the total profit by region. Sort the results from highest to lowest.

```python
# Your code here
df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
```

**Task 3.3:** Create a pivot table showing total sales and profit by Category and Sub-Category.

```python
# Your code here
df.pivot_table(index='Category', columns='Sub-Category', values=['Sales', 'Profit'], aggfunc='sum')
```

1. A school asserted that its students are more intelligent than the average student population. To test this, the IQ scores of 50 students were analyzed, resulting in an average score of 110. Given that the population mean IQ is 100 with a standard deviation of 15, determine whether the principal's claim is valid at a 5% significance level.


Null Hypothesis: x = 100
<br>
Alternative Hypothesis: x > 100

Test is a right-tailed test since the alternative uses > as the school claims its students have a higher average IQ.

For selecting the statistical test use a one-sample z-test for a population mean because:

The population standand deviation is known:
σ = 15

The sample size is:
n = 50

The sample mean is:
x = 110

The significance level is:
𝛼 = 0.05

```python
# -------------------------------------------------
# Step 1: Define the hypotheses and sample values
# -------------------------------------------------

population_mean = 100 #Hypothesized population mean,
population_std = 15 #Population standard deviation
sample_size = 50 #Sample size
sample_mean = 110 #Sample mean
alpha = 0.05 #Significance level

print("Null hypothesis: x = 100")
print("Alternative hypothesis: x > 100")
print("Test type: Right-tailed one-sample z-test")
```

```python
# The calculation for standard_error has been moved to the z-statistic calculation cell (LDkLSFjd21b8) for better dependency management and to resolve NameErrors.
```

Compare the result with the decision threshold:

For a right-tailed test at the 5% significance level, the critical z-value is: 1.645 because 4.71 > 1.645 the test statistic falls in the rejection region. The p-value is approximately: p < 0.00001
Since p < 0.05, reject the null hypothesis.

```python
import numpy as np # Added for robustness, as standard_error calculation is now here.

# -------------------------------------------------
# Step 2: Calculate the standard error (Moved from previous cell Hkaqzpbn2YrK)
# -------------------------------------------------
standard_error = population_std / np.sqrt(sample_size)
print("Standard error:", standard_error)

# -------------------------------------------------
# Step 3: Calculate the z-test statistic
# -------------------------------------------------

z_statistic = (sample_mean - population_mean) / standard_error
print("Z-statistic:", z_statistic)
```

```python
import scipy.stats as stats # Kept for robustness as confirmed necessary

# -------------------------------------------------
# Step 4: Calculate the p-value and critical value
# -------------------------------------------------

# Right-tailed p-value
p_value = 1 - stats.norm.cdf(z_statistic)
print("P-value:", p_value)

# Critical value
critical_value = stats.norm.ppf(1 - alpha)
print("Critical value:", critical_value)
```

State the conclusion
There is statistically significant evidence that the school's mean student IQ is greater than 100.

Therefore:
At the 5% significance level, the sample provides sufficient evidence to support the principal's claim that the school's students have a higher mean IQ than the general student population. However, the result is valid only if the 50 students were selected through a reasonably random, representative sampling procedure. The test supports a difference in the average IQ score, not the broader claim that every student is more intelligent.

```python
# -------------------------------------------------
# Step 5: Make the statistical decision
# -------------------------------------------------

if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

2. A company claims that the average response time for their customer support hotline is less than 5 minutes. You conducted a sample of 36 calls and found that the average response time was 4.7 minutes with a standard deviation of 1.2 minutes. Conduct a hypothesis test to determine whether there is sufficient evidence to support the company's claim.

### Hypotheses and Test Type (Problem 2)

**1. Define the Null and Alternative Hypotheses:**
*   **Null Hypothesis ($H_0$):** The average response time is greater than or equal to 5 minutes ($\mu \ge 5$).
*   **Alternative Hypothesis ($H_1$):** The average response time is less than 5 minutes ($\mu < 5$).

**2. Determine the Test Type:**
Since the alternative hypothesis states "less than" ($< 5$), this is a **left-tailed test**.

### Identifying the Statistical Test and Parameters (Problem 2)

**3. Identify the Statistical Test:**
We will use a **one-sample z-test** for a population mean because:
*   The sample size ($n = 36$) is greater than 30, allowing us to use the sample standard deviation as an estimate for the population standard deviation.

**Parameters:**
*   Population mean under null hypothesis ($\mu_0$): 5 minutes
*   Sample size ($n$): 36 calls
*   Sample mean ($\bar{x}$): 4.7 minutes
*   Sample standard deviation ($s$): 1.2 minutes
*   Significance level ($\alpha$): 0.05 (commonly used, not explicitly stated in the problem)

```python
# -------------------------------------------------
# Step 1: Define parameters for Problem 2
# -------------------------------------------------

population_mean_p2 = 5        # Hypothesized population mean
sample_size_p2 = 36          # Sample size
sample_mean_p2 = 4.7          # Sample mean
sample_std_p2 = 1.2           # Sample standard deviation
alpha_p2 = 0.05               # Significance level

print(f"Null hypothesis: mu >= {population_mean_p2}")
print(f"Alternative hypothesis: mu < {population_mean_p2}")
print("Test type: Left-tailed one-sample z-test")
```

### Calculating Test Statistics (Problem 2)

**4. Calculate the Standard Error, Z-statistic, P-value, and Critical Value:**

```python
import numpy as np
import scipy.stats as stats

# -------------------------------------------------
# Step 2: Calculate the standard error for Problem 2
# -------------------------------------------------
standard_error_p2 = sample_std_p2 / np.sqrt(sample_size_p2)
print("Standard error:", standard_error_p2)

# -------------------------------------------------
# Step 3: Calculate the z-test statistic for Problem 2
# -------------------------------------------------
z_statistic_p2 = (sample_mean_p2 - population_mean_p2) / standard_error_p2
print("Z-statistic:", z_statistic_p2)

# -------------------------------------------------
# Step 4: Calculate the p-value and critical value for Problem 2
# -------------------------------------------------

# Left-tailed p-value
p_value_p2 = stats.norm.cdf(z_statistic_p2)
print("P-value:", p_value_p2)

# Critical value for a left-tailed test
critical_value_p2 = stats.norm.ppf(alpha_p2)
print("Critical value:", critical_value_p2)
```

### Making a Statistical Decision and Conclusion (Problem 2)

**5. Make the Statistical Decision:**
Compare the p-value to the significance level, or the z-statistic to the critical value.

```python
# -------------------------------------------------
# Step 5: Make the statistical decision for Problem 2
# -------------------------------------------------

print(f"P-value: {p_value_p2}")
print(f"Significance level (alpha): {alpha_p2}")
print(f"Z-statistic: {z_statistic_p2}")
print(f"Critical value: {critical_value_p2}")

if p_value_p2 < alpha_p2:
    print("Decision: Reject the null hypothesis")
else:
    print("Decision: Fail to reject the null hypothesis")
```

### Conclusion (Problem 2)

**6. State the Conclusion:**

```python
# -------------------------------------------------
# Step 6: State the conclusion for Problem 2
# -------------------------------------------------

if p_value_p2 < alpha_p2:
    conclusion = (
        "There is sufficient evidence at the 0.05 significance level to support the company's claim "
        "that the average response time for their customer support hotline is less than 5 minutes."
    )
else:
    conclusion = (
        "There is not sufficient evidence at the 0.05 significance level to support the company's claim "
        "that the average response time for their customer support hotline is less than 5 minutes."
    )

print(conclusion)
```

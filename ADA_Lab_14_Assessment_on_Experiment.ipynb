##PART 1

Here are eight challenge scenarios that you can consider, identify variables:

1. Scenario: Students are investigating the impact of different study environments on memory retention.

Independent variable: Study environment
This is what the students change or manipulate. Examples might be: Quiet room, room with music, noisy/common area

Dependent variable: Memory retention
This is what they measure after studying, such as: Number of words remembered, quiz score, percentage of information correctly recalled.

Controlled variables: Factors kept as consistent as possible so they do not distort the results. Examples include: amount of study time, same material being studied, same delay before the memory test, same testing procedure, similar lighting, temperature, and instructions where appropriate.

Participants: The students are the subjects being studied.

Study environment -> Memory retention

2. Scenario: Researchers are exploring the effects of a new educational app on learning outcomes in a diverse group of students.

Independent variable: Use of the new educational app. For example: students who use the app vs. students who do not, or different amounts of app usage.

Dependent variable: Learning outcomes
This could be measured through: Test scores, Assignment grades, Knowledge retention, Skill improvement.

Possible control variables: Factors researchers mmay try to keep consistent or statistically account for: Amount of instructional time, course content, teacher/instructor, prior academic performance, access to learning resources.

Possible demographic/confounding variables: because the group is described as diverse, researchers may also examine: Age, Grade level, Socioeconomic background, language proficiency, prior technology experience, educational background.

The basic relationship is:

Educational app use -> Learning outcomes

3. Scenario: Investigating the influence of different smartphone notification settings on productivity.?

Independent variable: Smartphone notification setting. All notifications enabled, only priority notifications enabled, notifications silenced, do not disturb mode.

Dependent variable: productivity. Number of tasks completed, time required to complete a task, accuracy or error rate, amount of focused work time.

Controlled variables: Factors that should be kept as consistent as possible: Type and difficulty of the task, Amount of time given, work environment, phone model or app access, instructions given to participants.

Possible confounding variables:
Individual phone-use habits, motivation, baseline attention span, sleep/fatigue, familiarity wwith the task.

The core relationship is:
Notification settings -> Productivity

4. Scenario: Students are studying the impact of peer pressure on decision-making.

Independent variable: Level of peer pressure. No peer pressure, low peer pressure, high peer pressure, peers agreeing vs. disagreeing with the participant.

Dependent variable: decision-making. Choice making, willingness to take risks, accuracy of decisions, time taken to make a decision, whether the student changes an original answer to match the group.

Controlled variables: Same decision-making task, same intructions, same amount of time, same number of peers, same testing environment.

Possible confounding variables: Age, Personality, Self-confidence, relationship with peers, desire for social approval, previous experience with the situation.

The core relationship is:
Peer pressure -> decision-making



5. Scenario: Researchers are using surveys to assess the relationship between sleep patterns and academic performance.

Using surveys to assess a relationship rather than manipulating a variable, this is most naturally a correlational study.

Predictor variable: Sleep patterns. Hours of sleep per night, Bedtime/wake time, sleep consistency, sleep quality.

Outcome variable: Academic performance. GPA, exam scores, course grades, assignment performance.

Possible confounding/control variables: Study hours, course difficulty, prior academic ability, stress, employment/workload, health, age or grade level.

Sleep patterns -> Academic performance
The important distinction is that you generally would not call sleep the independent variable in the strict experimental sense because researchers aren't assigning students different amounts of sleep. They're measuring naturally occurring differences.

So the research question might be:
Is there a relationship between the amount of sleep students receive and their academic performance?

*correlation does not by itself establish causation.

6. Scenario: Investigating the relationship between different exercise routines and physical fitness.

Independent/predictor variable: Exercise routine: Cardio, Strength training, Combination training, high-intensity interval training, frequency or duration of exercise.

Dependent/outcome variable: Physical fitness. Cardiovascular endurance, strength, flexibility, body composition, fitness-test scores.

Possible control/confounding variables: Age, Sex, Diet, baseline fitness level, sleep, health status, exercise intensity and duration.

Exercise rountine -> Physical fitness
If researchers assign participants to different exercise routines, this can be an experimental study, with exercise routine as the  independent variable.

Observing people's existing exercise routines and compare them with fitness levels, it is correlational, so "predictor variable" and "outcome variable" are more precise terms.

7. Scenario: Researchers are examining the effects of a new teaching method on student performance, involving manipulation of teacher behavior.

The is an experimental scenario because the researchers are deliberately manipulating teacher behavior.

Independent variable: Teaching method/teacher behavior. Traditional teaching method, new teaching method, different levels or types of teacher interaction, feedback, or instructinal behavior.

Dependent variable: Student performance. Test scores, assignement grades, skill mastery, knowledge retention, course performance.

Controlled variables: same curriculum/content, same amount of instructional time, same assessments, similar classroom conditions, similar student grade level or ability where possible.

Possible confounding variables: teacher experience, student motivation, prior academic ability, class size, attendance, difference between schools or classrooms.

The core relationship is:

Manipulated teaching method -> Student performance.

Since the teacher behavior is intentionally changed, the teaching method is clearly the independent variable, while student performance is the dependent variable. This design can prrovide stronger evidence of causation than a purely correlational study, especially if students or classrooms are randomly assigned to conditions.

8. Scenario: Students are exploring the impact of environmental factors on mood changes.

Independent variable: Environmental factors. Lighting, Noise level, temperature, crowding, Natural vs. artificial surroundings, indoor vs. outdoor setting.

Dependent variable: Mood changes. Happiness, Stress, Anxiety, Calmness, Irritability, Overall mood score.

Possible control/confounding variables:
time of day, sleep, recent stressful events, caffeine use, social interactions, baseline mood, physical health.

Environmental factors -> Mood changes
If students deliberately change the environment and measure mood afterward, it is an experimental study.

If they simply observe existing environments and compare them with reported mood, it is correlational.

##PART 2

Conceptualize 2 experimental designs. Explain how you would conduct it in terms of:

1. What variables would it involved?
2. How would you gather the data?
3. How would you measure the variables?
4. How would you ensure a valid, unbiased, and informative results?

Experimental Design 1: Smartphone Notifications and Productivity.

Research question: Does reducing smartphone notifications improve productivity?

1. Variables involved
  -> Independent variable: notification setting
    Group 1: All notifications enabled.
    Group 2: Priority notifications only
    Group 3: Notifications completely silenced

    Dependent variable: Productivity

    Control variables: Same task, same time limit, same room, same instructions, similar device access.

    Possible confounders: Sleep, motivation, baseline phone dependence, attention ability.

  2. How to gather the data
    Randomly assign students to one of the three notifications conditions. Give everyone the same standardized academic or work task for a fixed period, such as 45 minutes. Record task performance during that period.
  
  3. How to measure the variables
    Notification condition is recorded categorically. Productivity could be measured using:
    Number of tasks completed.
    Perrcentage completed correctly
    Time per task
    Number of errors
    A combined productivity score could also be calculated.
  4. How to ensure valid, unbiased, informative results.
    Randomly assign participants to groups.
    Use identical tasks and instructions.
    Keep the environment consistent.
    Use objective performance measures rather than only self-reports.
    Use a sufficiently large sample.
    Do not tell participants the exact hypothesis, if ethically appropriate, to reduce expectancy efforts.
    Analyze whether differences between groups are statistically significant.


  Basic model:

  Notification setting -> Productivity

  Experimental Design 2: Environmental Conditions and Mood

  Research question: Does the physical environment influence students' mood?

  1. Variables involved
    Independent variable: Environmental condition
      Group 1: Quiet room with natural lighting
      Group 2: Bright artificial lighting with moderate noise
      Group 3: Dim lighting with background noise
    Dependent variable: Mood
    Control Variables: Time spent in room, temperature, activity performed, room size, intructions.
    Possible confounders: Previous mood, sleep, stress, caffiene, personality, recent life events.

  2. How to gater the data
    First, measure each participant's mood before entering the experimental condition. Randomly assign participants to one of the  environments. Have them remain in that environment for a standardized period, such as 30 minutes, while completing the same neutral activity. Measure mood again afterward.
  
  3. How to measure the variables
    Environmental condition is recorded categorically.
    Mood could be measured using a standardized mood questionnaire with numerical ratings for:
    Happiness, Stress, Calmness, Irritability, Positive and negative affect.
    The main outcome could be the change in mood score from before to after exposure.

  5. How to ensure valid, unbiased, informative results
  Random assignment reduces systematic differences betwen groups. Measure baseline mood before the experiment. Standardize temperature, duration, instructions, and activities. Use a validated mood scale. Keep the person scoring or analyzing results unaware of participant condition when possible. Recruit a diverse and sufficiently large sample. Compare both pre/post changes and differences across groups.

  Basic model:

  Environmental condition -> Change in mood

  The second design is especially strong because measuring mood before and after exposure lets researchers examine actual change rather than simply comparing different people's mood levels.

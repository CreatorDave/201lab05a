Bayes' Theorem problem. Here we are reversing a conditional probability.

What is known:

P(Red) = 0.25
P(Smooth | Red) = 0.80
P(Smooth) = 0.50

Desired outcome:
P(Red | Smooth)

Use Bayes' theorem:

P(Red | Smooth) = P(Smooth | Red) * P(Red)  / P(Smooth)

Substitute the values

P(Red | Smooth) = 0.80 X 0.25  / 0.50

= 0.20/0.50 = 0.40

Therefore,

P(Red | Smooth) = 0.40

The probability that a smooth marble is red is 40%

Although only 25% of all marbles are red, red marbles are especially likely to be smooth. Therefore, once I know that the selected marble is smooth, the probability that it is red rises from 25% to 40%.


```python
red = 0.25
smooth_red = 0.80
smooth = 0.50

red_smooth = smooth_red * red / smooth

print(red_smooth)
```
